# Chapter 1 - Tokenization and Bag of Words

## 1.1 Tokenization

### Step 1 - The corpus

We have 3 customer reviews and we want to know which 2 are similar

```python
docs = [
    "the discount was reduced",
    "we lowered our pricing",
    "the customer filed a complaint"
]
for d in docs:
    print(d)
```

**Corpus** - full collection of text we're working with

### Step 2 - Tokenization

```python
for d in docs:
    print(d.split())
```

Splitting text into words is called **tokenization** and each unit is a token

### Step 3 - Vocabulary

```python
vocab = sorted({w for d in docs for w in d.split()})
print(vocab)
```

```python
word_to_idx = {w: i for i, w in enumerate(vocab)}
print(word_to_idx)
```

### Step 4 - Bag of words

Now the document became 12 number, which makes the document comparable!!!

```python
import numpy as np

v = np.zeros(len(vocab))
print(v)
```

```python
doc = docs[0]

for w in doc.split():
    idx = word_to_idx[w]
    v[idx] += 1
    print(w," -> index", idx)

print(v)
```

```python
def bag_of_words(doc, vocab, word_to_idx):
    v = np.zeros(len(vocab))
    for w in doc.split():
        if w in word_to_idx:
            v[word_to_idx[w]] += 1
    return v
```

```python
X = np.array([bag_of_words(d, vocab, word_to_idx) for d in docs])
print(X)
print(X.shape)
```

**document-term matrix** - rows are documents and columns are terms.
 
In each row, most of it is zero -> **sparcity**

```python
docs2 = docs + ["the discount policy was the best discount"]

vocab2 = sorted({w for d in docs2 for w in d.split()})
word_to_idx2 = {w: i for i, w in enumerate(vocab2)}
```

```python
X2 = np.array([bag_of_words(d, vocab2, word_to_idx2) for d in docs2])
X2
```

```python
import pandas as pd
pd.DataFrame(X2, columns=vocab2, index = ["doc1", "doc2", "doc3", "doc4"])
```
